**Question 1 : What is Dimensionality Reduction? Why is it important in machine learning?**

**Answer 1:**

Dimensionality Reduction is the process of reducing the number of input features while preserving as much important information as possible.
It is important because:

  - It reduces computational cost and training time.

  - Helps avoid the curse of dimensionality.

  - Removes noise and redundant features.

  - Improves visualization of high-dimensional data (e.g., reducing to 2D or 3D).

  - Can improve model accuracy by reducing overfitting.


**Question 2: Name and briefly describe three common dimensionality reduction techniques..**

**Answer 2:**

**PCA (Principal Component Analysis)**

  - A linear method that transforms features into new uncorrelated components capturing maximum variance.

**t-SNE (t-Distributed Stochastic Neighbor Embedding)**

  - A non-linear technique used for visualization by preserving local structure in lower dimensions.

**LDA (Linear Discriminant Analysis)**

  - A supervised technique that maximizes class separability by finding linear combinations of features.


**Question 3: What is clustering in unsupervised learning? Mention three popular clustering algorithms.**

**Answer 3:**

Clustering is an unsupervised learning technique that groups similar data points based on patterns or similarities without predefined labels.

Three popular clustering algorithms:

  - K-Means

**Question 4: Explain the concept of anomaly detection and its significance.**

**Answer 4:**

Anomaly detection identifies data points that deviate significantly from the normal pattern.
It is significant because it helps detect:

  - Fraud (credit card transactions)

  - Network intrusions or cyber-attacks

  - Manufacturing defects

  - System failures and unusual patterns in time series

**Question 5: List and briefly describe three types of anomaly detection techniques.**

**Answer 5:**

**1. Statistical Methods**

  - Use probability distributions, mean, variance, z-scores to detect outliers.

**2. Machine Learning Methods**

  - Algorithms like Isolation Forest, One-Class SVM detect anomalies based on learned patterns.

**3. Deep Learning Methods**

  - Autoencoders detect anomalies by reconstruction error (anomalies reconstruct poorly).

**Question 6: What is time series analysis? Mention two key components of time series data.**

**Answer 6:**

Time series analysis is the study of data points collected over time to identify patterns, trends, and future values.

Two key components:

Trend: Long-term increasing or decreasing movement.

Seasonality: Regular repeating patterns (daily, monthly, yearly).

**Question 7: Describe the difference between seasonality and cyclic behavior in time series.**

**Answer 7:**

| **Seasonality**                                             | **Cyclic Behavior**                |
| ----------------------------------------------------------- | ---------------------------------- |
| Has fixed, known, repeating periods (e.g., every 12 months) | No fixed or predictable duration   |
| Driven by calendar/fixed events                             | Driven by economic/business cycles |
| Example: Higher sales every December                        | Example: Recession cycles          |


```python
'''
Question 8: Write Python code to perform K-means clustering on a sample dataset.

'''

# Answer 8:

import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Sample data
X = np.array([
    [1, 2], [1, 4], [1, 0],
    [10, 2], [10, 4], [10, 0]
])

# Perform K-means clustering
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(X)

# Cluster labels and centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

print("Cluster Labels:", labels)
print("Centroids:\n", centroids)

# Plotting
plt.scatter(X[:, 0], X[:, 1], c=labels)
plt.scatter(centroids[:, 0], centroids[:, 1], marker='X', s=200)
plt.title("K-Means Clustering")
plt.show()
```

```python
'''
Question 9: What is inheritance in OOP? Provide a simple example in Python.
'''

# Answer 9:
#   Inheritance allows a class (child class) to acquire properties and methods of another class (parent class).
#   It promotes code reusability and hierarchical structure.

class Animal:
    def sound(self):
        return "Some sound"

class Dog(Animal):
    def sound(self):
        return "Bark"

d = Dog()
print(d.sound())
```

**Question 10: How can time series analysis be used for anomaly detection?**

**Answer 10:**

Time series analysis detects anomalies by identifying data points that break the expected temporal patterns.

Techniques include:

  - Forecasting-based detection: Predict future values using ARIMA, LSTM; large prediction errors = anomalies.

  - Seasonal decomposition: Remove trend/seasonality and detect outliers in residuals.

  - Statistical thresholds: Outliers detected using z-score or moving averages.

  - Model-based techniques: Autoencoder, Prophet, Isolation Forest on time-series features.
