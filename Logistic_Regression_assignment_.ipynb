**Question 1 : What is Simple Linear Regression (SLR)? Explain its purpose.**

**Answer 1:**
 - Simple Linear Regression (SLR) is a statistical technique used to model the relationship between two variables — one independent variable (X) and one dependent variable (Y).
  - The purpose of SLR is to predict the value of Y based on X by fitting a straight line that best describes their relationship.


**Question 2: What are the key assumptions of Simple Linear Regression?**

**Answer 2:**
  - The main assumptions of SLR are:

    1. Linearity: The relationship between X and Y is linear.

    2. Independence: Observations are independent of each other.

    3. Homoscedasticity: The variance of residuals (errors) is constant across all levels of X.

    4. Normality: The residuals follow a normal distribution.

    5. No multicollinearity: Since there’s only one X in SLR, this assumption is automatically satisfied.

**Question 6: What is Logistic Regression? How does it differ from Linear Regression?**

**Answer 6:**
  - Logistic Regression is used to model the probability of a binary outcome (e.g., Yes/No, 0/1). It uses the logistic (sigmoid) function to map predicted values between 0 and 1.

  -  | Feature  | Linear Regression      | Logistic Regression       |
| -------- | ---------------------- | ------------------------- |
| Output   | Continuous             | Categorical (probability) |
| Function | Linear                 | Sigmoid (logistic)        |
| Use Case | Predict numeric values | Classify outcomes         |

```python
'''
Question 9: Write Python code to fit a simple linear regression model using scikit-learn
and print the slope and intercept.
(Include your Python code and output in the code box below.)
'''

# Answer 9:


import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression


X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Independent variable
Y = np.array([2, 4, 5, 4, 5])                 # Dependent variable


model = LinearRegression()
model.fit(X, Y)

# Print slope and intercept
print("Slope (β1):", model.coef_[0])
print("Intercept (β0):", model.intercept_)
```

**Question 10: How do you interpret the coefficients in a simple linear regression model?**

**Answer 10:**
  - Slope (β₁): Represents the change in Y for a one-unit increase in X.

    - Example: If β₁ = 0.7, it means Y increases by 0.7 units for every one-unit increase in X.

  - Intercept (β₀): Represents the value of Y when X = 0.
  - It’s where the regression line crosses the Y-axis.
